# Title:"Predicting Car Prices Using Support Vector Regression: A Machine Learning Approach"

#Problem Statement:
The automotive market is characterized by a vast array of vehicles with varying features, specifications, and price points, making it challenging for consumers to determine fair market values for cars. The objective of this project is to develop a predictive model using Support Vector Regression (SVR) to estimate the Manufacturer's Suggested Retail Price (MSRP) of cars based on various features such as engine specifications, fuel type, transmission type, and other relevant attributes.

By leveraging machine learning techniques, this project aims to:

Analyze and preprocess a dataset of car specifications and prices.
Implement and optimize a Support Vector Regression model to accurately predict car prices.
Evaluate the model's performance using metrics such as Mean Absolute Error (MAE), Mean Squared Error (MSE), and R-squared (R²).
Provide a user-friendly interface for users to input car specifications and receive MSRP predictions, thereby assisting potential buyers in making informed purchasing decisions.

#Objectives:
To understand the impact of different car features on pricing.
To apply data preprocessing techniques, including handling missing values and encoding categorical variables.
To utilize GridSearchCV for hyperparameter tuning of the SVR model.
To evaluate the model's performance and ensure its reliability for real-world applications.
This title and problem statement effectively summarize the goals of your project, the methodology used, and the significance of the work in the context of predicting car prices.






```python
import pandas as pd
from sklearn.model_selection import
train_test_split, GridSearchCV
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import mean_absolute_error,
mean_squared_error, r2_score
import numpy as np

# Step 1: Load your dataset
data = pd.read_csv('cars_data.csv')  # Adjust path if needed

# Step 2: Drop unnecessary columns (like Make, Model)
data = data.drop(columns=['Make', 'Model'], axis=1)

# Step 3: Handle missing values by dropping rows with missing data
data = data.dropna()

# Step 4: Encode categorical features (only for columns that are categorical)
categorical_cols = ['Engine Fuel Type',
                    'Transmission Type',
                    'Driven_Wheels',
                    'Market Category',
                    'Vehicle Size', 'Vehicle Style']

# Initialize LabelEncoder
le = LabelEncoder()

# Apply label encoding to categorical columns
for col in categorical_cols:
    data[col] = le.fit_transform(data[col])

# Step 5: Define features (X) and target (y)
X = data.drop('MSRP', axis=1)  # Features (excluding the target column)
y = data['MSRP']  # Target (car price)

# Step 6: Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 7: Split data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Step 8: Set up hyperparameter grid for GridSearchCV
param_grid = {
    'C': [0.1, 1, 10, 100],
    'epsilon': [0.01, 0.1, 0.2],
    'kernel': ['linear', 'rbf']
}

# Step 9: Perform GridSearchCV for optimal hyperparameters
grid_search = GridSearchCV(SVR(), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Step 10: Output the best parameters found during GridSearchCV
print(f"Best parameters: {grid_search.best_params_}")

# Step 11: Train the best model from GridSearchCV
svr_best = grid_search.best_estimator_
svr_best.fit(X_train, y_train)

# Step 12: Make predictions on the test set
y_pred = svr_best.predict(X_test)

# Step 13: Evaluate the model using different evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

# Step 14: Print evaluation metrics
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared (R²): {r2:.2f}")

# Step 15: User input for prediction
print("\nPlease enter the following details to predict the MSRP of a car:")
engine_fuel_type = input("Engine Fuel Type: ")
transmission_type = input("Transmission Type: ")
driven_wheels = input("Driven Wheels: ")
market_category = input("Market Category: ")
vehicle_size = input("Vehicle Size: ")
vehicle_style = input("Vehicle Style: ")
# Add other numerical features as needed
# Example:
# engine_hp = float(input("Engine HP: "))
# curb_weight = float(input("Curb Weight: "))
# etc.

# Prepare the input data for prediction
user_input = pd.DataFrame({
    'Engine Fuel Type': [engine_fuel_type],
    'Transmission Type': [transmission_type],
    'Driven_Wheels': [driven_wheels],
    'Market Category': [market_category],
    'Vehicle Size': [vehicle_size],
    'Vehicle Style': [vehicle_style],
    # Add other numerical features here
})

# Encode categorical features for user input
for col in categorical_cols:
    user_input[col] = le.transform(user_input[col])

# Scale the user input
user_input_scaled = scaler.transform(user_input)

# Make the prediction
predicted_msrp = svr_best.predict(user_input_scaled)

# Output the predicted MSRP
print(f"The predicted MSRP for the entered car details is: ${predicted_msrp[0]:.2f}")
```
